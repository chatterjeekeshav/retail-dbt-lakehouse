```python
from pyspark.sql import functions as F

# Create Bronze schema
spark.sql("""
CREATE SCHEMA IF NOT EXISTS retail_bronze
""")

# Read source tables
customers = spark.table("retail_lakehouse.src_customers")
products = spark.table("retail_lakehouse.src_products")
orders = spark.table("retail_lakehouse.src_orders")
order_items = spark.table("retail_lakehouse.src_order_items")

print("Source data loaded successfully.")
```

```python
# Write raw data to Bronze as Delta

customers.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_bronze.customers")

products.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_bronze.products")

orders.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_bronze.orders")

order_items.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("retail_bronze.order_items")

print("Bronze ingestion completed.")
```

```python
spark.sql("SHOW TABLES IN retail_bronze").show()
```
